```python
import numpy as np
import pandas as pd

# Roll the die 1000 times
rolls = np.random.randint(1, 7, 1000)

# Store in Pandas DataFrame
df = pd.DataFrame({"Roll": rolls})

# Numbers greater than 3
greater_than_3 = df[df["Roll"] > 3]

# Even numbers among numbers greater than 3
even_numbers = greater_than_3[greater_than_3["Roll"] % 2 == 0]

# Conditional probability
probability = len(even_numbers) / len(greater_than_3)

print("Total rolls:", len(df))
print("Numbers greater than 3:", len(greater_than_3))
print("Even numbers and greater than 3:", len(even_numbers))
print("Conditional Probability:", probability)
```
