```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy
import seaborn as sns
from scipy import stats

np.random.seed(42)

# Parameters
n, p = 20, 0.35

# Binomial distribution
binom = stats.binom(n, p)

# Values of X
x = np.arange(0, n + 1)

# PMF: Probability Mass Function
pmf = binom.pmf(x)

# CDF: Cumulative Distribution Function
cdf = binom.cdf(x)

# Mean and moments
mean = binom.mean()
moment_1 = binom.moment(1)
moment_2 = binom.moment(2)
moment_3 = binom.moment(3)
moment_4 = binom.moment(4)

# Variance and standard deviation
variance = binom.var()
std_dev = binom.std()

# Skewness and kurtosis
skewness = binom.stats(moments="s")
kurtosis = binom.stats(moments="k")
# Probability of exactly 7 successes
probability = binom.pmf(7)


# Display statistics
print("--- Moments and Statistics ---")
print("Mean E[X] =", mean)
print("First moment E[X] =", moment_1)
print("Second moment E[X²] =", moment_2)
print("Third moment E[X³] =", moment_3)
print("Fourth moment E[X⁴] =", moment_4)
print("Variance =", variance)
print("Standard Deviation =", std_dev)
print("Skewness =", skewness)
print("Excess Kurtosis =", kurtosis)
print("P(X = 7) =", probability)

# Probability distribution table
df = pd.DataFrame({
    "X": x,
    "PMF P(X=x)": pmf,
    "CDF P(X<=x)": cdf
})

print("\n--- Probability Distribution ---")
print(df.round(4).to_string(index=False))

# Plot PMF
plt.figure(figsize=(10, 5))
sns.barplot(x=x, y=pmf, color="steelblue")
plt.title("Binomial PMF (n=20, p=0.35)")
plt.xlabel("X")
plt.ylabel("P(X=x)")
plt.tight_layout()
plt.show()

# Plot CDF
plt.figure(figsize=(10, 5))
plt.step(x, cdf, where="post")
plt.title("Binomial CDF (n=20, p=0.35)")
plt.xlabel("X")
plt.ylabel("P(X<=x)")
plt.ylim(0, 1.05)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
